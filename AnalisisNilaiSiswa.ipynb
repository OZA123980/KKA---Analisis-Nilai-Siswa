```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Membaca dataset
df = pd.read_csv('student_dropout_behavior_dataset.csv')

# Melihat 5 data pertama
print("======= 5 DATA PERTAMA ======")
print(df.head())

# Informasi dataset
print("======= INFORMASI DATASET =====")
df.info()

# Mengecek data kosong
print("======= DATA KOSONG ======")
print(df.isnull().sum())

# Mengecek data duplikat
print("======= DATA DUPLIKAT =====")
print(df.duplicated().sum())

# Statistik deskriptif awal
print("======= STATISTIK DESKRIPTIF =======")
print(df.describe())

# Menghapus kolom yang seluruh datanya kosong
df = df.dropna(axis=1, how='all')

# Menghapus data duplikat
df = df.drop_duplicates()

print("===== DATA SETELAH PEMBERSIHAN =======")
print(df.shape)
print(df.isnull().sum())
```

Dataset berhasil dimuat, diperiksa, dan kolom assignments_submitted yang seluruhnya kosong dihapus.

TAHAP 2 – Analisis Statistik
Tujuan Kami menghitung:
1.Rata-rata nilai Quiz.
2.Rata-rata nilai UTS.
3.Rata-rata nilai UAS.
4.Rata-rata GPA sebelumnya.
5.Persentase kehadiran kuliah.
6.Persentase kehadiran laboratorium.

```python
# 1. Rata-rata nilai Quiz
rata_quiz = df[
    ['quiz1_marks', 'quiz2_marks', 'quiz3_marks']
].mean().mean()

# 2. Rata-rata nilai UTS
rata_uts = df['midterm_marks'].mean()

# 3. Rata-rata nilai UAS
rata_uas = df['final_marks'].mean()

# 4. Rata-rata GPA
rata_gpa = df['previous_gpa'].mean()

# 5. Persentase kehadiran kuliah
kehadiran = (
    df['lectures_attended'].sum() /
    df['total_lectures'].sum()
) * 100

# 6. Persentase kehadiran laboratorium
kehadiran_lab = (
    df['labs_attended'].sum() /
    df['total_lab_sessions'].sum()
) * 100

# Menampilkan hasil
print("===== HASIL ANALISIS STATISTIK =====")

print("Rata-rata Quiz:", round(rata_quiz, 2))
print("Rata-rata UTS:", round(rata_uts, 2))
print("Rata-rata UAS:", round(rata_uas, 2))
print("Rata-rata GPA:", round(rata_gpa, 2))
print("Kehadiran Kuliah:", round(kehadiran, 2), "%")
print("Kehadiran Laboratorium:", round(kehadiran_lab, 2), "%")
```

Interpretasi:Hasil ini digunakan untuk mengetahui gambaran performa akademik, hasil ujian,prestasi sebelumnya,serta tingkat kehadiran siswa.

TAHAP 4 – Deteksi Outlier
Kita menggunakan metode IQR (Interquartile Range) untuk mendeteksi nilai yang terlalu rendah atau terlalu tinggi dibandingkan sebagian besar data.

```python
# Menghitung Q1 dan Q3
Q1 = df['final_marks'].quantile(0.25)
Q3 = df['final_marks'].quantile(0.75)

# Menghitung IQR
IQR = Q3 - Q1

# Menentukan batas bawah dan atas
batas_bawah = Q1 - (1.5 * IQR)
batas_atas = Q3 + (1.5 * IQR)

# Mencari outlier
outlier = df[
    (df['final_marks'] < batas_bawah) |
    (df['final_marks'] > batas_atas)
]

print("===== HASIL DETEKSI OUTLIER =====")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Batas bawah:", batas_bawah)
print("Batas atas:", batas_atas)

print("\nJumlah outlier:", len(outlier))
print(outlier[['student_id', 'name', 'final_marks']])

# Visualisasi outlier
plt.figure(figsize=(8, 5))
plt.boxplot(df['final_marks'])

plt.title('Deteksi Outlier Nilai UAS')
plt.ylabel('Nilai UAS')
plt.show()
```

Interprestasi:Outlier merupakan data yang berada di luar batas normal berdasarkan metode IQR.Nilai yang terdeteksi perlu diperiksa lebih lanjut untuk mengetahui apakah disebabkan oleh kondisi tertentu atau kesalahan pencatatan.

TAHAP 5 – Membuat Dashboard
Sesuai ketentuan guru, dashboard harus menggabungkan minimal 4 grafik dalam satu figure menggunakan subplot.

Kita gabungkan:
1.Bar Chart nilai Quiz.
2.Line Chart nilai evaluasi.
3.Scatter Plot kehadiran dan nilai UAS.
4.Histogram nilai UAS.

```python
fig, ax = plt.subplots(2, 2, figsize=(14, 10))

# Grafik 1: Bar Chart
nilai_quiz = df[
    ['quiz1_marks', 'quiz2_marks', 'quiz3_marks']
].mean()

ax[0, 0].bar(nilai_quiz.index, nilai_quiz.values)
ax[0, 0].set_title('Rata-rata Nilai Quiz')
ax[0, 0].set_xlabel('Jenis Quiz')
ax[0, 0].set_ylabel('Nilai')

# Grafik 2: Line Chart
nama = ['Quiz 1', 'Quiz 2', 'Quiz 3', 'UTS', 'UAS']

nilai = [
    df['quiz1_marks'].mean(),
    df['quiz2_marks'].mean(),
    df['quiz3_marks'].mean(),
    df['midterm_marks'].mean(),
    df['final_marks'].mean()
]

ax[0, 1].plot(nama, nilai, marker='o')
ax[0, 1].set_title('Rata-rata Nilai Evaluasi')
ax[0, 1].set_xlabel('Evaluasi')
ax[0, 1].set_ylabel('Nilai')

# Grafik 3: Scatter Plot
ax[1, 0].scatter(
    df['lectures_attended'],
    df['final_marks']
)

ax[1, 0].set_title('Kehadiran vs Nilai UAS')
ax[1, 0].set_xlabel('Kehadiran')
ax[1, 0].set_ylabel('Nilai UAS')

# Grafik 4: Histogram
ax[1, 1].hist(
    df['final_marks'],
    bins=10,
    edgecolor='black'
)

ax[1, 1].set_title('Distribusi Nilai UAS')
ax[1, 1].set_xlabel('Nilai UAS')
ax[1, 1].set_ylabel('Frekuensi')

# Judul dashboard
fig.suptitle(
    'DASHBOARD ANALISIS PERFORMA SISWA',
    fontsize=16
)

plt.tight_layout()
plt.show()
```
